- Find the average age (sum of all items divided by their number )
- Find the range of the ages (max - min)
- Compare the value of (min - average) and (max - average), use **[abs()]** method

### 1

````python
ages = [19, 23, 19, 25, 21, 20, 25, 26, 25, 24]

# Sort
ages.sort()
print(ages)

# Min and Max
min_age = min(ages)
max_age = max(ages)

print("Min:", min_age)
print("Max:", max_age)
````

````python
ages.append(min_age)
ages.append(max_age)

print(ages)
````

````python
middle = len(ages) // 2

median = (ages[middle - 1] + ages[middle]) / 2

print("Median:", median)
````

````python
average = sum(ages) / len(ages)

print("Average:", average)
````

````python
age_range = max(ages) - min(ages)

print("Range:", age_range)
````

````python
min_difference = abs(min_age - average)
max_difference = abs(max_age - average)

print(min_difference)
print(max_difference)
````

````python
middle = len(countries) // 2

print(countries[middle])
````

### 3

````python
middle = (len(countries) + 1) // 2

first_half = countries[:middle]
second_half = countries[middle:]

print(first_half)
print(second_half)
````

### 4

````python
country1, country2, country3, *scandic_countries = countries

print(country1)
print(country2)
print(country3)
print(scandic_countries)
````
